```python
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor

# 1. Prepare Features and Target
target_column = df['price']
features = df.drop('price', axis=1)

# 2. Split into Training and Test Sets
X_train, X_test, y_train, y_test = train_test_split(
    features, target_column, test_size=0.18, random_state=42
)

print('Training set shape:', X_train.shape)
print('Test set shape:', X_test.shape)

# 3. Initialize Models
lin_reg = LinearRegression()
tree_reg = DecisionTreeRegressor(random_state=42)
forest_reg = RandomForestRegressor(random_state=42)

# 4. Train Models on Training Data
lin_reg.fit(X_train, y_train)
tree_reg.fit(X_train, y_train)
forest_reg.fit(X_train, y_train)

# 5. Predict on Test Set (evaluating on unseen data)
lin_preds = lin_reg.predict(X_test)
tree_preds = tree_reg.predict(X_test)
forest_preds = forest_reg.predict(X_test)

# 6. Calculate Test RMSE
lin_rmse = root_mean_squared_error(y_test, lin_preds)
tree_rmse = root_mean_squared_error(y_test, tree_preds)
forest_rmse = root_mean_squared_error(y_test, forest_preds)

print('\nTest Performance:')
print('Linear Regression RMSE:', lin_rmse)
print('Decision Tree RMSE:', tree_rmse)
print('Random Forest RMSE:', forest_rmse)
```
